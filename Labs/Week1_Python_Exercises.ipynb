```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 100 values between -pi and pi
# np.pi is a built-in constant for 3.1415..
x = np.linspace(-np.pi, np.pi, 100)

# Compute the y values using the sine function
y = np.sin(x)

# Plot the graph with a specific color (green)
plt.plot(x, y, color='green')
plt.title("Sine Wave")
plt.show()

# Scatter plot with dots (purple)
plt.scatter(x, y, color='purple')
plt.show()
```

```python
def fibonacci(N):
    # Start the list with the first two Fibonacci numbers
    sequence = [0, 1]
    
    # Loop to add numbers until the list is length N
    while len(sequence) < N:
        # Sum the last two elements
        next_val = sequence[-1] + sequence[-2]
        sequence.append(next_val)
        
    return sequence[:N] # Return only the first N numbers

# Test it
print(fibonacci(67))
```

```python
def bubble(values):
    n = len(values)
    # Loop through the entire list
    for i in range(n):
        # Last i elements are already in place
        for j in range(0, n - i - 1):
            # Swap if the element found is greater than the next element
            if values[j] > values[j + 1]:
                values[j], values[j + 1] = values[j + 1], values[j]

# Test it
my_list = [64, 34, 25, 12, 22, 11, 90]
bubble(my_list)
print("Sorted list:", my_list)
```

```python
#EXTENSION TASK
import numpy as np
import matplotlib.pyplot as plt

# Generate the same 100 random numbers from Exercise 3
uniform_data = np.random.rand(100)
normal_data = np.random.randn(100)

# Produce a histogram showing the two distributions
# 'alpha' makes the bars transparent so you can see where they overlap
plt.hist(uniform_data, bins=15, alpha=0.5, label='Uniform', color='blue')
plt.hist(normal_data, bins=15, alpha=0.5, label='Normal', color='orange')

plt.title("Histogram of Distributions")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend() # Shows the labels in the corner
plt.show()
```
